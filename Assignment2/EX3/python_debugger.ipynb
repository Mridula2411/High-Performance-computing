### Task 3.1 Reflection

#### ADVANTAGES OF USING pdb

- Python debugger is extremely helpful in understanding just what is going on in a program while its being executed. A debugger can be useful for identifying logical errors that are not immediately visible from the output as it allows a step-by-step execution of a program, making it easier to understand how the code flows and how variables change over time. 

- It has commands which can show, which line of code its about to run, with an arror, and the part of code where the current execution is halted due to the breakpoint.

- Another big advantage is the ability to inspect variables at any point in the execution of the code. Using pdb, we are not only able to inspect the variables, but we are also able to understand their change and whether the logic in the code is behaving as expected.

- More control is available while running the program, such as stepping into a function, skipping a certain number of lines, or running the program until a predetermined point. This makes debugging a program more systematic, clear, and efficient compared to using only print statements.

- We can also enable or disable all breakpoints before the program runs.

- When stuck, help offers all the list of commands and aliases which helps to not remember all the commands and their shorthand.

- When the debugger quits enexpectedly, there is a stack trace and logging by default which is very helpful.


#### CHALLENGES OF USING pdb

- The disadvantage with pdb is that everything works in the terminal, and to some extent, it can be quite tricky at first, especially because the commands are short and lack descriptive words.

- Another problem is associated with stepping through large loops, which may be time-consuming. In some programs, particularly those used to calculate mathematical results, stepping through large loops may be tedious.
